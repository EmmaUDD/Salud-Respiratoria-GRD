### 2024

```python
df_2024 = pd.read_csv('datos_grd/GRD_PUBLICO_2024.txt', 
                      sep='|', 
                      encoding='Latin-1',
                      low_memory=False)

#parse_dates=['FECHA_NACIMIENTO', 'FECHA_INGRESO', 'FECHATRASLADO1', 'FECHATRASLADO2', 'FECHATRASLADO3', 'FECHATRASLADO4', 'FECHATRASLADO5', 'FECHATRASLADO6', 'FECHATRASLADO7', 'FECHATRASLADO8', 'FECHATRASLADO9', 'FECHAALTA', 'FECHAPROCEDIMIENTO1', 'FECHAINTERV1']
```

```python
df_2024.head()
```

```python
parse_dates = ['FECHA_NACIMIENTO', 'FECHA_INGRESO', 'FECHATRASLADO1', 'FECHATRASLADO2', 'FECHATRASLADO3', 'FECHATRASLADO4', 'FECHATRASLADO5', 'FECHATRASLADO6', 'FECHATRASLADO7', 'FECHATRASLADO8', 'FECHATRASLADO9', 'FECHAALTA', 'FECHAPROCEDIMIENTO1', 'FECHAINTERV1']

for col in parse_dates:
    df_2024[col] = pd.to_datetime(df_2024[col], format='%Y-%m-%d', errors='coerce')

df_2024['FECHA_NACIMIENTO'].dropna(inplace=True)
df_2024['FECHA_INGRESO'].dropna(inplace=True)
```

```python
limpiar_cols = ['SEXO', 'COMUNA', 'TIPO_INGRESO', 'PREVISION']

for col in limpiar_cols:
    df_2024 = df_2024[df_2024[col] != 'DESCONOCIDO']
```

```python
df_2024['SEXO'].value_counts()
```

```python
df_2024['COMUNA'].value_counts()
```

```python
df_2024['TIPO_INGRESO'].value_counts()
```

```python
df_2024['PREVISION'].value_counts()
```

```python
df_2024['ETNIA'].value_counts()
```

```python
df_2024.to_csv('datos_unidos.csv', index=False, encoding='utf-8-sig')
```

```python
del df_2024
gc.collect()
```
